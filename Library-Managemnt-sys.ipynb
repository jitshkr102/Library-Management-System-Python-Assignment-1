```python
# ============================================================
# LIBRARY MANAGEMENT SYSTEM
# ============================================================

# Main dictionary to store all books currently available.
# Key   = Book ID (integer)
# Value = Dictionary containing book title and author.


books = {
    101: {"title": "Python", "author": "Raj"},
    102: {"title": "SQL", "author": "Amit"},
    103: {"title": "Java", "author": "Ravi"},
    104: {"title": "C++", "author": "Kiran"},
    105: {"title": "Comics", "author": "Anil"},
    106: {"title": "Science", "author": "Priya"},
    107: {"title": "History", "author": "Arun"},
    108: {"title": "Maths", "author": "Neha"},
    109: {"title": "Stories", "author": "Rahul"},
    110: {"title": "Poetry", "author": "Sneha"}

    }


# issued_books = {} : stores the student id as dictionary key, student name, book id, books name, books author, 
# used lists is multiple books is issued to particular students
issued_books= {}

# available_books = {} just for keep the records for all the available books
available_books = {}

# deleted_books = {} this is used to keep the records for which books are being deleted from the library
deleted_books = {}

# keeping the record of book id. by default 110. So, the next book will be added to 111
book_id = 110

# Using while True: repeating untill the user decides to exit 
while True:
    print("\n==== LIBRARY MANAGEMENT SYSTEM ====")
    print("Press 1: Add a new book to the library")
    print("Press 2: Display all available books")
    print("Press 3: Search for books")
    print("Press 4: Update book details")
    print("Press 5: Delete a book from a library")
    print("Press 6: Issue a book to the student")
    print("Press 7: Return an issued book")
    print("Press 8: Display only available books")
    print("Press 9: Display the issued books")
    print("Press 10: Exit")

# Enter options to perform the managing work
    options = input("Enter a key to Operate")

    # Adding New book to the library. A new book id is generated by incrementing the id. 
    # After successful addition of books , Details of the book is printed.
    if options == "1":
        book_id = book_id + 1
        title = input("Enter Book Title")
        Author = input("Enter Author Name")
    
        books[book_id] = {"title":title,
                          "author":Author}
        
        print("Book Details Added Successfully")
        
        print("Book ID:", book_id)
        print("Book Title:", books[book_id]["title"])
        print("Author Name:", books[book_id]["author"])
        
     # displaying the all books which is in the library    
    elif options == "2":
        
        if books == {}: #here checking the books availability. if any book is present in the library
            print("No available Books")
        else:
            for book_id in books:
                print("\n Book Id:", book_id)
                print("Book Title:", books[book_id]["title"])
                print("Author Name: ",books[book_id]["author"])
    
    # Search the books 
                
    elif options == "3":
        print("==== Search Books ====")
        print("Press 1: Search Books by ID")
        print("Press 2: Search Books by Name")
        print("Press 3: Search Books by Author")
        
        id = input("Enter key to search")
        
        # Search books by ID
        if id == "1":
            search_id = int(input("Enter Book Id"))
    
            if search_id in books:
                print("Book is available")
                print("Book Id:", search_id)
                print("Book Name:", books[search_id]["title"])
                print("Author:", books[search_id]["author"])
            else:
                print("Book is not Available")
                
         # search books by book name   
        elif id == "2":
            search_name = input("Enter Book Name").lower()
            found = False
    
            for book_id, book_title in books.items():
                if search_name in book_title["title"].lower():
                    print("Book Id:", book_id)
                    print("Book Name:", book_title["title"])
                    print("Author:", book_title["author"])
                    found = True
            if not found:
                print("Book is not available")
                
        # search books by author name
        elif id == "3":
            search_author = input("Enter Author Name").lower()
            found = False
    
            for book_id, book_author in books.items():
                if search_author in book_author["author"].lower():
                    print("Book Id:", book_id)
                    print("Book Name:", book_author["title"])
                    print("Author:", book_author["author"])
                    found = True
            if not found:
                print("Book is not available")
        else:
            print("Invalid search Option")
            
    # Update Books: This option allows to update the details of the book. Like author name, title .
    elif options == "4":
        print("==== Update Books ====")
        
        update_key = int(input("Enter Book Id"))
        if update_key in books:
            print("Book Found")
            print("Current Book Name",books[update_key]["title"])
            print("Author Name",books[update_key]["author"])
            
            print("\nPress 1: Update Book Name")
            print("Press 2: Update Book Author")

            update_options = input("Enter option to update")
            if update_options == "1":
                
                books[update_key]["title"] = input("Enter the name")
                print("Book Name has been updated")

            elif update_options == "2":
                books[update_key]["author"] = input("Enter Author Name")
                print("Book Author has been updated")
            else:
                print("Invalid Option")
        else:
            print("Book Not found")
                
    elif options == "5":
        # del_book is for the which book is going to be deleted. 
        # created a seperated dictionary for deleted books that store the details regarding the same.
        del_book = int(input("Enter Book ID: "))

        if del_book not in books:
            print("Book not found")
        else:
            deleted_books[del_book] = books[del_book] #copying the details of the books which is going to be deleted. to a new dictionary
            
            print("Book Id", del_book)
            print("Book Name:", books[del_book]["title"])
            print("Book Author:", books[del_book]["author"])
            
            del books[del_book]
            print("Books Deleted successfully")


    # Issue a book to a student.
    # this options student gives a name of the book then by name the book is searched.
    # then form that book id is retrieved
    # issued books are updated in new dictionary issued_books which keeps the details of the student and book details
    # student details such as student_id and Student_name
    # book details such as book_id, book_name, book_author.
    # used lists in dictionary for is someone issues multiple books
    
    elif options == "6":
        issue_book = input("Enter Book Name").lower()

        found = False

        for book_id, book_title in books.items():
            if issue_book in book_title["title"].lower():
                found = True
                issue_book_id = book_id
                student_id = input("enter the student Id ")

                if student_id in issued_books:
                    issued_books[student_id]["issued_book"].append({"book":issue_book_id,
                                         "book_name":books[issue_book_id]["title"],
                                         "book_author":books[issue_book_id]["author"]
                                         })
                else:
                    student_name = input("Enter the student Name")

                    issued_books[student_id] = {"student_name":student_name,
                              "issued_book":[{"book":issue_book_id,
                                             "book_name":books[issue_book_id]["title"],
                                             "book_author":books[issue_book_id]["author"]
                                             }]}
                del books[issue_book_id]
                print(f"Book issued to: {issued_books[student_id]['student_name']}")
                break
                
        if not found:
            print("Book is not available")
        
    # return the books
    # student return the books than it is updated in the book dictionary
    # The returned book is removed from the student's issued-book list.
    # If the student has no remaining issued books, the student
    # record is removed from the issued_books dictionary.

    elif options == "7":
        student_return_id = input("Enter the student Id")
        
        if student_return_id not in issued_books:
            print("Student Id not found")
        else:
            
            student_books = issued_books[student_return_id]["issued_book"]
    
            print("-- Issued Books --")
            print("Student Id:", student_return_id)        
            print("Student Name:", issued_books[student_return_id]["student_name"])        
            print(f"Issued Books: {student_books}")
    
            return_book_id = int(input("Enter the book id:"))
            for item in student_books:
                if item["book"] == return_book_id:
                    books[return_book_id] = { "title": item["book_name"],
                                              "author": item["book_author"]
                            
                    }      
                    student_books.remove(item)
                    print("Book returned Successfully")
                    break
            if not student_books:
                del issued_books[student_return_id]
                print("No more books issued to this student.")

    # OPTION 8: DISPLAY AVAILABLE BOOKS

    elif options == "8":
        if books == {}:
            print("No available Books")
        else:
            for book_id in books:
                print("\n Book Id:", book_id)
                print("Book Title:", books[book_id]["title"])
                print("Author Name: ",books[book_id]["author"])

    
        # This option displays all books currently issued to students.
        # Used loops to get the details of the books issued from issued_books dictionary and displayed here
        # The Book ID, book name, and author name are displayed
        # for each issued book.  
        
    elif options == "9":
        if not issued_books:
            print("No books have been issued")
        else:
            print("=== Issued Books ===")
            
            for student_id , student_details in issued_books.items():
                for item in student_details["issued_book"]:
                    print("Book ID:", item["book"])
                    print("Book Name:", item["book_name"])
                    print("Book Author:", item["book_author"])
                    
        
    elif options == "10":
        print("Exiting Library Management System")
        break
        
        
```
